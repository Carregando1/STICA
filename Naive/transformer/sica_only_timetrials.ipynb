Part 2: Testing Dim/Num Timesteps against Time Required for SICA Simulation: Trial 1: Dim Testing

```python
#Requires  bit less than 2 hours of runtime.
data: List[Tuple] = []
iter = 11
def generateicsrt(dim, num_steps, example = 0):
    ic = np.random.choice([0,1], size=(dim, dim))
    srt = np.random.choice([0,1], size=(num_steps, dim, dim, 18))
    srtcomp = np.zeros((num_steps, dim, dim, 18, 4))
    for i in range(num_steps):
        for j in range(dim):
            for k in range(dim):
                for l in range(18):
                    srtcomp[i][j][k][l] = np.array([8-l%9, l%9, int(np.floor(l/9)), srt[i][j][k][l]])
    return ic, srtcomp
for i in [10, 20, 32, 50, 64]:
    for j in [50, 100, 200, 500]:
        print(f"OPERATING TIMESTEPDIM: {i} step {j}")
        for _ in range(iter):
            stepdim = [i,j]
            icsrt = generateicsrt(stepdim[0], stepdim[1])
            ic = icsrt[0]
            srt = icsrt[1]
            time_init = time.time()
            state = CurrentState(initial=ic, rules=srt)
            res = state.generate()
            data.append(((i, j),time.time()-time_init))
            print(f"Iter {_} completed in {time.time()-time_init}s")


```

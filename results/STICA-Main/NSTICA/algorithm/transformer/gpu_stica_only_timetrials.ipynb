Part 2: Testing Dim/Num Timesteps against Time Required for STICA Simulation: Trial 1: Dim Testing

```python
#Requires  bit less than 2 hours of runtime.
data: List[Tuple] = []
iter = 11
def generateicsrt(dim, num_steps, example = 0):
    ic = np.random.choice([0,1], size=(dim, dim))
    srt = np.random.choice([0,1], size=(num_steps, dim, dim, 18))
    return ic, srt
for i in [10, 20, 32, 50, 64, 100, 150]:
    for j in [50, 100, 200, 500, 1000]:
        print(f"OPERATING TIMESTEPDIM: {i} step {j}")
        for _ in range(iter):
            stepdim = [i,j]
            icsrt = generateicsrt(stepdim[0], stepdim[1])
            ic = icsrt[0]
            srt = icsrt[1]
            time_init = time.time()
            res = transition(ic, srt, j, 2, None)
            data.append(((i, j),time.time()-time_init))
            print(f"Iter {_} completed in {time.time()-time_init}s")


```

```python
#Results for n-state STICA only: non-compatible with other results.
dataincomp: List[Tuple] = []
iter = 11
def generateicsrt(dim, num_steps, states, example = 0):
    ic = np.random.choice(np.arange(0, states), size=(dim, dim))
    srt = np.random.choice(np.arange(0, states), size=(num_steps, dim, dim, states * math.comb(7+states, 8)))
    return ic, srt
conds = [(50, 10, 3), (50, 20, 3), (50, 32, 3), (50, 64, 3), (100, 10, 3), (100, 20, 3), (100, 32, 3), (200, 10, 3), (200, 20, 3), (200, 32, 3), (500, 10, 3), (500, 20, 3), (50, 10, 4), (50, 20, 4), (50, 32, 4), (100, 10, 4), (100, 20, 4), (200, 10, 4), (500, 10, 4), (50, 10, 5), (50, 20, 5), (100, 10, 5), (200, 10, 5), (50, 10, 6)]
for cond in conds:
    print(f"OPERATING TIMESTEPDIM: {cond[1]} step {cond[0]} state {cond[2]}")
    for _ in range(iter):
        stepdim = [cond[1], cond[0], cond[2]]
        icsrt = generateicsrt(cond[1], cond[0], cond[2])
        ic = icsrt[0]
        srt = icsrt[1]
        time_init = time.time()
        res = transition(ic, srt, cond[0], cond[2], None)
        dataincomp.append(((cond[1], cond[0], cond[2]),time.time()-time_init))
        print(f"Iter {_} completed in {time.time()-time_init}s")


```

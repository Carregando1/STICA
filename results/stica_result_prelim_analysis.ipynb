Organizing and Filtering Data

```python
def condify(gs, its):
    res = []
    for it in its:
        for g in gs:
            res.append((it, g))
    if (calculate): print(res)
    return res
lowgs = condify([10, 20, 32, 64], [50, 100, 200, 500])
highgs = condify([10, 20, 32, 64, 100], [50, 100, 200, 500])
"""In the following three sets for STICAOnly, the order of It and GS is reversed."""
lowsticags = condify([50, 100, 200, 500], [10, 20, 32, 50, 64])
medsticags = condify([50, 100, 200, 500, 1000], [10, 20, 32, 50, 64, 100])
highsticags = condify([50, 100, 200, 500, 1000], [10, 20, 32, 50, 64, 100, 150])
```

```python
#Organize data into dictionary form.

#STICA only conds = lowsticags for Naive, medsticags for CPU, and highsticags for GPU.
def condize_stica(stica):
    assert len(stica)/11 % 1 == 0, "11 its per cond"
    res = {}
    for i in range(int(len(stica)/11)):
        res[stica[i*11][0]] = [b[1] for b in stica[i*11:i*11+11]]
    if (calculate): print(res)
    return res
#conds = highgs for 2S SICA2S, lowgs for other 2S or comp gens, and nstica_conds for NSTICA
#itcond = 10 for Naive, 20 for GPU, 50 for CPU
def condize_gen(gen, conds, itcond):
    assert len(gen)/itcond == len(conds), f"Wrong cond or itcond (gen length: {len(gen)} ≠ cond length: {len(conds)} * its: {itcond}). For NSTICA_Gen modify procedure"
    res = {}
    for i in range(int(len(gen)/itcond)):
        res[conds[i]] = gen[i*itcond:i*itcond+itcond]
    if (calculate): print(res)
    return res

#Special organization for NSTICA because of the 2 trials in 50It-64GS-4S and 8 trials in 200It-20GS-6S, making a total of 3110 timelogs instead of the expected 3200.
def condize_nstica_gen(gen, conds, itcond):
    assert len(gen) == 3110, f"Wrong timelog set."
    buf = 0
    res = {}
    for i in range(64):
        if (conds[i] == (50, 64, 4)):
            res[conds[i]] = gen[i*itcond+buf:i*itcond+buf+2]
            buf = -48
            continue
        res[conds[i]] = gen[i*itcond+buf:i*itcond+itcond+buf]
    if (calculate): print(res)
    return res

nvstc = condize_stica(naive_stica)
st2sstc = condize_stica(stica_2s_stica)
st2sgstc = condize_stica(stica_2s_gpu_stica)
nststc = condize_stica(nstica_stica)
nstgstc = condize_stica(nstica_gpu_stica)
nststincmpc = condize_stica(nstica_stica_incomp)
nstgstincmpc = condize_stica(nstica_gpu_stica_incomp)

nvgnc = condize_gen(naive_gen, lowgs, 10)
st2sgc = condize_gen(stica_2s_gen, highgs, 50)
st2sggc = condize_gen(stica_2s_gpu_gen, highgs, 20)
st2scgc = condize_gen(stica_2s_comp_gen, lowgs, 50)
st2sgcgc = condize_gen(stica_2s_gpu_comp_gen[400:], lowgs, 20)
nstgc = condize_nstica_gen(nstica_gen, nstica_conds, 50)
nstggc = condize_gen(nstica_gpu_gen, nstica_gpu_conds, 20)
nstcgc = condize_gen(nstica_comp_gen, lowgs, 50)
nstgcgc = condize_gen(nstica_gpu_comp_gen[880:], lowgs, 20)
```

```python
#Filter out negative values.
def filter(dict):
    for key in dict:
        dict[key] = np.array(dict[key])[np.array(dict[key])>=0]
    if (calculate): print(dict)
    return dict
nvstc = filter(nvstc)
st2sstc = filter(st2sstc)
st2sgstc = filter(st2sgstc)
nststc = filter(nststc)
nstgstc = filter(nstgstc)
nststincmpc = filter(nststincmpc)
nstgstincmpc = filter(nstgstincmpc)

nvgnc = filter(nvgnc)
st2sgc = filter(st2sgc)
st2sggc = filter(st2sggc)
st2scgc = filter(st2scgc)
st2sgcgc = filter(st2sgcgc)
nstgc = filter(nstgc)
nstggc = filter(nstggc)
nstcgc = filter(nstcgc)
nstgcgc = filter(nstgcgc)

"""In dictionaries, the key is (It, GS, States) for gen and (GS, It, States) for STICA-Only."""
```

```python
#Filter out first-run outliers in the STICA-Only set. Only is done if f-variables are True.
fStica = True
fGen = True
def remove(dict):
    for key in dict:
        dict[key] = dict[key][1:]
    if (calculate): print(dict)
    return dict
if (fStica):
    nvstc = remove(nvstc)
    st2sstc = remove(st2sstc)
    st2sgstc = remove(st2sgstc)
    nststc = remove(nststc)
    nstgstc = remove(nstgstc)
    nststincmpc = remove(nststincmpc)
    nstgstincmpc = remove(nstgstincmpc)
if (fGen):
    nvgnc = remove(nvgnc)
    st2sgc = remove(st2sgc)
    st2sggc = remove(st2sggc)
    st2scgc = remove(st2scgc)
    st2sgcgc = remove(st2sgcgc)
    nstgc = remove(nstgc)
    nstggc = remove(nstggc)
    nstcgc = remove(nstcgc)
    nstgcgc = remove(nstgcgc)
```

Analysis- STICA Only- Non Incompatible (1 = SICA2S GPU Runtime, <1 = faster, >1 = slower)

```python
#Quotient of two means with standard deviation?
"""Order of runtimes: Naive, STICA2S, STICA2S GPU, NSTICA, NSTICA GPU"""
stica2s = {}
for cond in highsticags:
    runs = []
    if cond in nvstc: runs.append(np.mean(nvstc[cond]))
    else: runs.append(math.inf)
    if cond in st2sstc: runs.append(np.mean(st2sstc[cond]))
    else: runs.append(math.inf)
    runs.append(np.mean(st2sgstc[cond]))
    if cond in nststc: runs.append(np.mean(nststc[cond]))
    else: runs.append(math.inf)
    if cond in nstgstc: runs.append(np.mean(nstgstc[cond]))
    else: runs.append(math.inf)
    runs /= np.mean(st2sgstc[cond])
    if (calculate or finc): print(f"{cond} matrix: {runs}")
    stica2s[cond] = runs
```

Analysis- STICA Only- Incompatible (1 = NSTICA Runtime)

```python
nstica_incmp_conds = [(50, 10, 3), (50, 20, 3), (50, 32, 3), (50, 64, 3), (100, 10, 3), (100, 20, 3), (100, 32, 3), (200, 10, 3), (200, 20, 3), (200, 32, 3), (500, 10, 3), (500, 20, 3), (50, 10, 4), (50, 20, 4), (50, 32, 4), (100, 10, 4), (100, 20, 4), (200, 10, 4), (500, 10, 4), (50, 10, 5), (50, 20, 5), (100, 10, 5), (200, 10, 5), (50, 10, 6)]
#The conds for both incmp are the same.
"""Order of runtimes: NSTICA, NSTICA GPU"""
nstica = {}
for cond in nstica_incmp_conds:
    runs = []
    runs.append(np.mean(nststincmpc[(cond[1], cond[0], cond[2])]))
    if (cond[1], cond[0], cond[2]) in nstgstincmpc: runs.append(np.mean(nstgstincmpc[(cond[1], cond[0], cond[2])]))
    else: runs.append(math.inf)
    runs /= np.mean(nststincmpc[(cond[1], cond[0], cond[2])])
    if (calculate or finc): print(f"{(cond[1], cond[0], cond[2])} matrix: {runs}")
    nstica[(cond[1], cond[0], cond[2])] = runs
```

Analysis- Genetic Algorithm- 2 States, Naive-Comparable (1 = STICA2S Comparable Runtime)

```python
"""Order of runtimes: Naive, STICA2S, STICA2S GPU, NSTICA, NSTICA GPU"""
stica2sgen = {}
for cond in lowgs:
    runs = []
    if cond in nvgnc: runs.append(np.mean(nvgnc[cond]))
    else: runs.append(math.inf)
    runs.append(np.mean(st2scgc[cond]))
    if cond in st2sgcgc: runs.append(np.mean(st2sgcgc[cond]))
    else: runs.append(math.inf)
    if cond in nstcgc: runs.append(np.mean(nstcgc[cond]))
    else: runs.append(math.inf)
    if cond in nstgcgc: runs.append(np.mean(nstgcgc[cond]))
    else: runs.append(math.inf)
    runs /= np.mean(st2scgc[cond])
    if (calculate or finc): print(f"{cond} matrix: {runs}")
    stica2sgen[cond] = runs
```

Analysis- Genetic Algorithm- 2 States, Fixed # of Mutations (GS in IC, GS^2 in SRT) (1 = STICA2S Runtime)

```python
"""Order of runtimes: STICA2S, STICA2S GPU, NSTICA, NSTICA GPU"""
stica2sgenfixed = {}
for cond in highgs:
    runs = []
    runs.append(np.mean(st2sgc[cond]))
    if cond in st2sggc: runs.append(np.mean(st2sggc[cond]))
    else: runs.append(math.inf)
    #2S Conds are written with three arguments within NSTICA.
    if (cond[0], cond[1], 2) in nstgc: runs.append(np.mean(nstgc[(cond[0], cond[1], 2)]))
    else: runs.append(math.inf)
    if (cond[0], cond[1], 2) in nstggc: runs.append(np.mean(nstggc[(cond[0], cond[1], 2)]))
    else: runs.append(math.inf)
    runs /= np.mean(st2sgc[cond])
    if (calculate or finc): print(f"{cond} matrix: {runs}")
    stica2sgenfixed[cond] = runs
```

Analysis- Genetic Algorithm- N>2 States (1 = NSTICA Runtime)

```python
nsticagen = {}
#nstica_conds contains all testcases of nstica_gpu_conds.
for cond in nstica_conds:
    runs = []
    runs.append(np.mean(nstgc[cond]))
    if cond in nstggc: runs.append(np.mean(nstggc[cond]))
    else: runs.append(math.inf)
    runs /= np.mean(nstgc[cond])
    if (calculate or finc): print(f"{cond} matrix: {runs}")
    nsticagen[cond] = runs
```
